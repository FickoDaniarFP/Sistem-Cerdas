1. Buktikan dengan tabel kebenaran bahwa ¬(p ∧ q) ekuivalen dengan ¬p ∨ ¬q (hukum De Morgan) — tunjukkan kolom keduanya identik pada keempat baris.

#### Bukti Hukum De Morgan: ¬(p ∧ q) ≡ ¬p ∨ ¬q

| $p$ | $q$ | $p \land q$ | $\neg(p \land q)$ | $\neg p$ | $\neg q$ | $\neg p \lor \neg q$ |
| :-: | :-: | :-: | :-: | :-: | :-: | :-: |
| T | T | T | **F** | F | F | **F** |
| T | F | F | **T** | F | T | **T** |
| F | T | F | **T** | T | F | **T** |
| F | F | F | **T** | T | T | **T** |

> **Kesimpulan:** Kolom `¬(p ∧ q)` dan `¬p ∨ ¬q` menghasilkan kombinasi nilai kebenaran yang identik pada keempat baris (F, T, T, T). Hal ini membuktikan bahwa kedua bentuk tersebut ekuivalen secara logis.

5. Kasus penguin: penguin is-a burung tetapi tidak bisa terbang. Usulkan cara representasi frame menangani pengecualian ini, dan jelaskan pelajaran umumnya tentang pengetahuan dunia nyata.



> Pada struktur representasi pengetahuan, sistem *frame* mengendalikan kondisi pengecualian menggunakan mekanisme **penimpaan nilai bawaan (*default value override*)** pada slot tertentu di tingkat subkelas.
>
> **Frame Aves / Burung (Kelas Induk / General):**
> * `is-a`: Hewan
> * `Penutup_Tubuh`: Bulu
> * `Bisa_Terbang` *(default)*: Ya
>
> **Frame Penguin (Kelas Turunan / Spesifik):**
> * `is-a`: Burung
> * `Cara_Bergerak`: Berenang
> * `Bisa_Terbang` *(override)*: Tidak
>
> **Pelajaran Umum tentang Pengetahuan Dunia Nyata:** Pengetahuan di dunia nyata tidak bersifat mutlak melainkan berpola *non-monotonic reasoning* (penalaran non-monotonik). Artinya, sebuah aturan umum dapat dibatalkan ketika ditemukan bukti khusus baru. Sistem berbasis pengetahuan harus menggunakan nilai *default* demi efisiensi pemrosesan awal, tetapi wajib menyediakan slot khusus untuk menimpa (*override*) aturan umum tersebut saat menangani kasus-kasus anomali atau pengecualian.

Kegiatan 1 — NumPy: Operasi Vektor

```python
import numpy as np

uts = np.array([78, 90, 66, 85, 72])
uas = np.array([80, 94, 52, 88, 75])

nilai_akhir = 0.4 * uts + 0.6 * uas
print("Nilai akhir :", nilai_akhir)
print("Rata-rata   :", nilai_akhir.mean())
print("Tertinggi   :", nilai_akhir.max())
print("Terendah    :", nilai_akhir.min())
```

Pengembangan 1

```python
# Pengembangan: Menambahkan bobot tugas, penentuan status kelulusan (np.where), dan penghitungan jumlah mahasiswa lulus (np.sum)
tugas = np.array([85, 92, 70, 88, 80])

nilai_akhir_total = (0.3 * uts) + (0.5 * uas) + (0.2 * tugas)
status = np.where(nilai_akhir_total >= 70, "LULUS", "GAGAL")

print("Nilai akhir total :", np.round(nilai_akhir_total, 1))
print("Status kelulusan  :", status)
print("Rata-rata total   :", np.round(nilai_akhir_total.mean(), 2))
print("Jumlah lulus      :", np.sum(status == "LULUS"), "mahasiswa")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Membaca dataset QS World University Rankings 2025
df_qs = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/qs-world-rankings-2025.csv")
# Konversi kolom ke numerik untuk mencegah galat tipe data
cols = ["Academic Reputation", "Employer Reputation", "Citations per Faculty"]
for col in cols:
    df_qs[col] = pd.to_numeric(df_qs[col], errors="coerce")
```

```python
# 1. Histogram: Sebaran Academic Reputation
plt.figure(figsize=(6, 4))
plt.hist(df_qs["Academic Reputation"].dropna(), bins=8, color="#2b5c8f", edgecolor="white")
plt.title("Sebaran Academic Reputation Score")
plt.xlabel("Academic Reputation Score")
plt.ylabel("Jumlah Universitas")
plt.show()
```

```python
# 2. Scatter Plot: Hubungan Academic Reputation dan Employer Reputation
plt.figure(figsize=(6, 4))
plt.scatter(df_qs["Academic Reputation"], df_qs["Employer Reputation"], color="#e65100", alpha=0.5, s=15)
plt.title("Hubungan Academic vs Employer Reputation")
plt.xlabel("Academic Reputation")
plt.ylabel("Employer Reputation")
plt.show()
```

```python
# 3. Diagram Batang: Rata-rata Academic Reputation Berdasarkan Size (Baris 2-4)
plt.figure(figsize=(6, 4))
rata_size = df_qs.groupby("Size")["Academic Reputation"].mean().reset_index().iloc[1:4]
plt.bar(rata_size["Size"], rata_size["Academic Reputation"], color=["#43a047", "#f57c00", "#1e88e5"])
plt.title("Rata-rata Academic Reputation Berdasarkan Size (Baris 2-4)")
plt.xlabel("Size Kampus")
plt.ylabel("Rata-rata Score")
plt.show()
```

```python
# 4. Boxplot (TAMBAHAN DIAGRAM BARU): Distribusi Citations per Faculty di 4 Negara Teratas
plt.figure(figsize=(6, 4))
top_4_negara = df_qs["Location Full"].value_counts().head(4).index
df_top = df_qs[df_qs["Location Full"].isin(top_4_negara)]

df_top.boxplot(column="Citations per Faculty", by="Location Full", grid=False)
plt.title("Distribusi Citations per Faculty di 4 Negara Teratas")
plt.suptitle("") # Menghapus judul bawaan pandas
plt.xlabel("Negara")
plt.ylabel("Citations per Faculty Score")
plt.xticks(rotation=15)
plt.show()
```
